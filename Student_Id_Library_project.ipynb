# ***Task 2:***

```python
df = pd.read_csv("/content/Task1_combined_data.csv")
```

```python
print("Shape of the data:", df.shape)
```

```python
print("Data info:", df.info())
```

```python
print("Number of null cells before dealing with them:", df.isna().sum())
```

```python
df['return_date'] = df['return_date'].fillna('Not returned yet')
```

```python
df['grade'] = df['grade'].fillna('Unknown grade')
```

```python
df['join_date'] = df['join_date'].fillna('Unknown joining date')
```

```python
print("Number of null cells after dealing with them:", df.isna().sum())
```

```python
print("Number of duplicated rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print("Shape of data after removing duplicates:", df.shape)
```

```python
df["membership_status"] = df["membership_status"].str.strip()
```

```python
df["membership_status"] = df["membership_status"].str.title()
```

```python
df["neighborhood"] = df["neighborhood"].str.strip()
```

```python
df["neighborhood"] = df["neighborhood"].str.title()
```

```python
print(df['grade'].value_counts())
```

```python
print(df[df['return_date'] < df['checkout_date']])
```

```python
df.to_csv("Task2_cleaned_data.csv", index=False)
```

# Data cleaning report:

1- Dealing with missing values: Missing values in return dates were replaced with "Not returned yet" and missing values in grades and join dates were replaced with "Unknown".

2- Dealing with duplicates: 8 duplicated rows were found and dropped.

3- Text formatting: Extra spaces were cleaned using strip() function and the text in neighborhood and membership_status was edited to be the same thing using strip() and title().

4- Data integrity: Grades, checkout and return dates were verified that they are valid and logical.
