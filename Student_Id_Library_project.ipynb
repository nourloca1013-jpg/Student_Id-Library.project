Part 1:

```python
import numpy as np
import pandas as pd
import sqlite3 as sql
```

```python
conn = sql.connect("/content/library.db.db")
```

```python
# Question 1:
q1 = """SELECT member_id, COUNT(*)
FROM checkouts
GROUP BY member_id;"""
pd.read_sql_query(q1, conn)
```

```python
# Question 2:
q2 = """SELECT *
FROM books
WHERE author LIKE 'A%';"""
pd.read_sql_query(q2, conn)
```

```python
# Question 3:
q3 = """SELECT book_id, COUNT(*)
FROM checkouts
GROUP BY book_id
ORDER BY COUNT(*) DESC LIMIT 5;"""
pd.read_sql_query(q3, conn)
```

```python
# Question 4:
q4 = """SELECT member_id, COUNT(*)
FROM checkouts
GROUP BY member_id
ORDER BY COUNT(*) DESC LIMIT 10;"""
pd.read_sql_query(q4, conn)
```

```python
# Question 5:
q5 = """SELECT member_id, COUNT(*)
FROM checkouts
GROUP BY member_id
ORDER BY COUNT(*) ASC LIMIT 10;"""
pd.read_sql_query(q5, conn)
```

Part 2:

```python
query = "SELECT * FROM checkouts, members, books WHERE checkouts.member_id = members.member_id AND checkouts.book_id = books.book_id"
final_df = pd.read_sql(query, conn)

final_df.to_csv("Task1_combined_data.csv", index=False)
```
